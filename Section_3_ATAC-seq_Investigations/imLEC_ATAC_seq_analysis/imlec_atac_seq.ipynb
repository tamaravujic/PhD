## Workflow Steps
1. **Setup**: Login to HPC and navigate to data directory
2. **Environment**: Create and activate Conda environment with required tools
3. **Quality Control**: Process FASTQ files and assess data quality
4. **Read Processing**: Alignment, filtering, and peak calling
5. **Peak Analysis**: Generate chromatin accessibility profiles

## Prerequisites
- FASTQ files
- ATAC.yml environment file (conda environment specification)

## Required Tools
The conda environment includes:
- FastQC: quality control
- Bowtie2: read alignment
- Samtools: BAM file processing
- MACS2: peak calling
- And other ATAC-seq analysis tools

## Output
- Aligned BAM files
- Peak files (narrowPeak/broadPeak format)
- Quality control reports
- Chromatin accessibility signals
